```python

import os
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

OUTPUT_PATH = os.path.join("/Users/kgothalangwesi/Downloads", "relay_vs_confirmation.png")

fig, ax = plt.subplots(figsize=(9.5, 4.6), dpi=200)

#  Time axis in minutes, 0 to 70 
t = np.linspace(0.001, 70, 2000)

k_of_t = t / 10.0
reorg_risk = np.exp(-0.85 * k_of_t)

ax2 = ax.twinx()
ax2.plot(t, reorg_risk * 100, color="#b5303a", lw=2.2,
         label="Illustrative reorg risk vs depth $k$")
ax2.set_ylabel("Illustrative probability payment is\nreversed by a reorg (%)", color="#b5303a")
ax2.tick_params(axis="y", colors="#b5303a")
ax2.set_ylim(0, 105)

# Block arrivals every ~10 minutes
for i in range(1, 8):
    ax.axvline(i * 10, color="#888888", lw=0.8, ls="--", zorder=1)
    ax.text(i * 10, 1.05, f"Block\n(k={i})", ha="center", va="bottom",
             fontsize=7.5, color="#444444")


relay_box = mpatches.FancyBboxPatch((0, 0), 0.6, 1.0, boxstyle="round,pad=0.02",
                                     linewidth=0, facecolor="#2f6fb0", alpha=0.85, zorder=3)
ax.add_patch(relay_box)
ax.annotate("Compact-block relay window\n\u2248 a few seconds/hop)",
            xy=(0.6, 0.85), xytext=(6, 0.9),
            fontsize=8, color="#1c3f63",
            arrowprops=dict(arrowstyle="->", color="#2f6fb0", lw=1.2))

# Confirmation policy band: k=6 / ~60 min 
ax.axvspan(55, 65, color="#3f9142", alpha=0.18, zorder=0)
ax.annotate("Policy: settle R10\u2009000 remittance\nat k=6 (\u224860 min))",
            xy=(60, 0.5), xytext=(36, 0.55),
            fontsize=8, color="#235a27",
            arrowprops=dict(arrowstyle="->", color="#3f9142", lw=1.2))

# Axes / title / legend
ax.set_xlim(0, 70)
ax.set_ylim(0, 1.15)
ax.set_yticks([])
ax.set_title(
    "Block relay time vs confirmation policy for a R10\u2009000 remittance\n",
    fontsize=9.7,
)

lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(lines2, labels2, loc="center right", bbox_to_anchor=(1.0, 0.65),
          fontsize=8, frameon=False)

fig.tight_layout()
fig.savefig(OUTPUT_PATH, facecolor="white")
print(f"Saved diagram to: {OUTPUT_PATH}")
```
